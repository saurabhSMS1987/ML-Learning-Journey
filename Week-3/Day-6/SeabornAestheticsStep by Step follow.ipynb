# Customizing Seaborn Aesthetics

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# "white", "dark", "whitegrid", "darkgrid", "ticks"
```

```python
tips = sns.load_dataset("tips")
```

```python
sns.set_style("white")
sns.scatterplot(data=tips, x="total_bill", y="tip")
```

```python
sns.set_style("whitegrid")
sns.scatterplot(data=tips, x="total_bill", y="tip")
```

```python
sns.set_style("dark")
sns.scatterplot(data=tips, x="total_bill", y="tip")
```

```python
sns.set_style("darkgrid")
sns.scatterplot(data=tips, x="total_bill", y="tip")
```

```python
sns.set_style("ticks")
sns.scatterplot(data=tips, x="total_bill", y="tip")
```

```python
sns.lineplot(data=tips, x="total_bill", y="tip")
```

```python
sns.set_theme(style="darkgrid")
```

```python
sns.lineplot(data=tips, x="total_bill", y="tip")
```

```python
sns.axes_style()
```

```python
sns.lineplot(data=tips, x="total_bill", y="tip")
```

```python
sns.set_style({'axes.facecolor': '#FC427B', 'grid.color': '#9AECDB'})
```

```python
sns.lineplot(data=tips, x="total_bill", y="tip")
```

```python
sns.set_style("ticks", {'axes.facecolor': '#FC427B', 'grid.color': '#9AECDB'})
```

```python
sns.scatterplot(data=tips, x="total_bill", y="tip")
```

```python
sns.set_style("darkgrid")
```

```python
sns.scatterplot(data=tips, x="total_bill", y="tip")
```

## Custom Palettes

```python
sns.color_palette("pastel", 6)
```

```python
codes = ["#B33771", "#3B3B98", "#FD7272", "#9AECDB", "#D6A2E8"]
custom_pal = sns.color_palette(codes)
```

```python
sns.barplot(data=tips, x="day", y="total_bill", palette=custom_pal)
```

```python
sns.light_palette("#eb2f06")
```

```python
sns.dark_palette("#eb2f06")
```

```python
diamonds = sns.load_dataset("diamonds")
```

```python
sns.histplot(
    diamonds,
    x="price", hue="cut",
    multiple="stack",
    log_scale=True,
    palette=sns.dark_palette("#eb2f06", 5)
)
```

```python
sns.histplot(
    diamonds,
    x="price", hue="cut",
    multiple="stack",
    log_scale=True,
    palette="dark:olive"
)
```
